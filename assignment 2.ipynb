```python
#Program to find the largest of three numbers 


n1 = int(input("enter first no:"))
n2 = int(input("enter second no:"))
n3 = int(input("enter third no:"))

if (n1 >= n2) and (n1 >= n3):
    largest = n1
elif (n2 >= n1) and (n2 >= n3):
    largest = n2
else:
    largest = n3
print ("largest no is:", largest )    
```
